```python
import mcvs
import pandas as pd
import numpy as np
```

```python
# Load the mCVs feature data:
mCVs_features = pd.read_parquet(mcvs.get_mCVs_path().replace('.csv', '_features.parquet'))
```

```python
# Compute the mean of all features:
means = np.mean(mCVs_features[mCVs_features.columns[3:]], axis=0) # first three columns are not features
```

```python
# Create test objects:
# An object at the center of the mCVs distribution:
totaly_a_mcv = pd.DataFrame([means], columns=mCVs_features.columns[3:])
totaly_a_mcv['objectId'] = 'totaly_a_mcv'
# Other objects far from the center:
really_not_a_mcv1 = pd.DataFrame([means + means*1000], columns=mCVs_features.columns[3:])
really_not_a_mcv1['objectId'] = 'really_not_a_mcv1'
really_not_a_mcv2 = pd.DataFrame([means - means*1000], columns=mCVs_features.columns[3:])
really_not_a_mcv2['objectId'] = 'really_not_a_mcv2'
really_not_a_mcv3 = pd.DataFrame([means + means*1500], columns=mCVs_features.columns[3:])
really_not_a_mcv3['objectId'] = 'really_not_a_mcv3'

# Concatenate in a single DataFrame to be inputted into `mcvs.eval_candidates`:
test_objects = pd.concat([totaly_a_mcv, really_not_a_mcv1, really_not_a_mcv2, really_not_a_mcv3], ignore_index=True)
```

```python
# Evaluate the scores of the test objects:
candidates = mcvs.eval_candidates(test_objects, score_threshold=0) # `score_threshold=0` allows to return all
candidates
# 'totaly_a_mcv' should have the maximum score and other objects a score of zero.
```

```python
# The maximum score should be equal to the number of mCVs in the feature set:
assert len(mCVs_features) == candidates['score'].max()
```

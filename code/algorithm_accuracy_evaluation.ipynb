##### Imports & setup

```python
import pandas as pd
import matplotlib.pyplot as plt
from magcvs_library.plots import explore_params
import warnings
# Disabling FutureWarnings:
warnings.filterwarnings('ignore', category=FutureWarning)
```

## Get data

```python
# The following paths should be changed accordingly. The data can be saved from the features_from_light_curves.ipynb notebook.
positive_g = pd.read_parquet('../../data/magcvs/features_g.parquet')
polars_g = pd.read_parquet('../../data/polars/features_g.parquet')
negative_g = pd.read_parquet('../../data/negative/features_g.parquet')

positive_r = pd.read_parquet('../../data/magcvs/features_r.parquet')
polars_r = pd.read_parquet('../../data/polars/features_r.parquet')
negative_r = pd.read_parquet('../../data/negative/features_r.parquet')

feature_names = positive_g.columns[3:]

# Adding labels for each class:
positive_g['class'] = 'positive'
polars_g['class'] = 'positive'
negative_g['class'] = 'negative'

positive_r['class'] = 'positive'
polars_r['class'] = 'positive'
negative_r['class'] = 'negative'
```

## Plots

### g filter

#### All mCVs, all features

```python
explore_params(positive=positive_g, negative=negative_g, N=1000)
plt.savefig('../figures/accuracy_plots/accuracy_all_g.svg')
```

#### All mCVs, no periodogram features

```python
periodogram_features = [
 'period_0',
 'period_s_to_n_0',
 'period_1',
 'period_s_to_n_1',
 'period_2',
 'period_s_to_n_2',
 'periodogram_amplitude',
 'periodogram_beyond_1_std',
 'periodogram_beyond_2_std',
 'periodogram_cusum',
 'periodogram_eta',
 'periodogram_inter_percentile_range_25',
 'periodogram_standard_deviation',
 'periodogram_percent_amplitude']
```

```python
explore_params(positive=positive_g, negative=negative_g, N=1000, feature_names=feature_names[~feature_names.isin(periodogram_features)])
plt.savefig('../figures/accuracy_plots/accuracy_no_periodogram_features_g.svg')
```

#### Only polars

```python
explore_params(positive=polars_g, negative=negative_g, N=1000)
plt.savefig('../figures/accuracy_plots/accuracy_polars_g.svg')
```

### r filter

#### All mCVs

```python
explore_params(positive=positive_r, negative=negative_r, N=1000)
plt.savefig('../figures/accuracy_plots/accuracy_all_r.svg')
```

#### Only polars

```python
explore_params(positive=polars_r, negative=negative_r, N=1000)
plt.savefig('../figures/accuracy_plots/accuracy_polars_r.svg')
```
